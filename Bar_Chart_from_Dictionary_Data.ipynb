```
import matplotlib.pyplot as plt

data = {
    "Invited Lectures/ Talk": 8,
    "Conference Paper Presented": 18,
    "FDP/Workshop/SSTP Attended/Training": 16,
    "Guest Editor (Special Issues SCI/SCIE/Scopus)": 5,
    "Special Session in Conference": 5,
    "Session Chair in Conference": 7,
}

# Extract activity names and counts
activities = list(data.keys())
counts = list(data.values())

# Define a list of colors for the bars
#colors = ['skyblue', 'lightgreen', 'lightcoral', 'lightsalmon', 'lightcyan', 'lightyellow']
colors = ['paleturquoise', 'salmon', 'orange', 'sandybrown', 'aquamarine', 'oldlace']
# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(activities, counts, color=colors)  # Use the list of colors

# Add labels and title
plt.xlabel("Activity", fontsize=14)  # Set font size 14
plt.ylabel("Count", fontsize=14)
plt.title("Total Counts\n", fontsize=16)  # Increased title font size
plt.xticks(rotation=45, ha="right", fontsize=12)  # Set x-axis label font size
plt.yticks(fontsize=12) # set y axis label font size

# Add gridlines
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels above the bars
for i, count in enumerate(counts):
    plt.text(activities[i], count + 0.5, str(count), ha='center', va='bottom', fontsize=14)  # Set font size 14

# Show the plot
plt.tight_layout()
plt.show()
```
