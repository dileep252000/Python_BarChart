```
import matplotlib.pyplot as plt

# Data
labels = ['Journals', 'Conferences', 'Books', 'Book Chapters', 'Patents', 'Article']
sizes = [32, 30, 6, 10, 9, 1]
colors = ['paleturquoise', 'salmon', 'orange', 'sandybrown', 'aquamarine', 'oldlace']

# Function to show numbers
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct * total / 100.0))
        return f'{val:d}'
    return my_autopct

# Create pie chart
plt.figure(figsize=(10, 10))
wedges, texts, autotexts = plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct=make_autopct(sizes),
    startangle=140,
    wedgeprops={'edgecolor': 'white'},
    textprops={'fontsize': 14, 'weight': 'bold'}
)

# Make label texts bold and bigger
for text in texts:
    text.set_fontsize(14)
    text.set_weight('bold')

# Make number texts bold and bigger
for autotext in autotexts:
    autotext.set_fontsize(14)
    autotext.set_weight('bold')

plt.title('Total Publication Count = 88\n\n', fontsize=16, weight='bold')
plt.axis('equal')

plt.show()
```
